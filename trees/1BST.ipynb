# Binary search tree - A Binary Search Tree (BST) is a specialized binary tree data structure that keeps its elements in a sorted order to enable fast searching, insertion, and deletion.
# All Values in left sub tress < parent node
# All values in right sub tree >= parent
# Tarversal means visiting everty node atleast once 

1. Inorder - Leftsubtree -> right subtree (This rule is followed by every sub tree) 
2. Preorder - root first , root -> left sub tree -> right sub tree 
3. Postorder - root last , left subtree -> right subtree -> root 


```python

class Node:
    def __init__(self, value):
        self.left = None
        self.data = value
        self.right = None


class BST:
    def __init__(self):
        self.root = None

    def insert(self, new_node):
        if self.root is None:  # BST is empty
            self.root = new_node
        else:
            temp = self.root

            while True:
                if new_node.data < temp.data:
                    if temp.left is None:
                        temp.left = new_node
                        return
                    else:
                        temp = temp.left

                else:
                    if temp.right is None:
                        temp.right = new_node
                        return
                    else:
                        temp = temp.right

    def inorder(self, rt):
        if rt is None:
            return
        else:
            self.inorder(rt.left)
            print(rt.data, end=" ")
            self.inorder(rt.right)


b1 = BST()

b1.insert(Node(100))
b1.insert(Node(10))
b1.insert(Node(450))
b1.insert(Node(0))
b1.insert(Node(65))
b1.insert(Node(300))

b1.inorder(b1.root)
```
