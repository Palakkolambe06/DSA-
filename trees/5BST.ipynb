LEVELWISE PRINTING

```python

class Node:
    def __init__(self, data):
        self.data = data
        self.left = None
        self.right = None

class BST:
    def __init__(self):
        self.root = None

    # INSERTION (Recursive)
    def insert(self, node):
        self.root = self._insert(self.root, node)

    def _insert(self, root, node):
        if root is None:
            return node

        if node.data < root.data:
            root.left = self._insert(root.left, node)

        elif node.data > root.data:
            root.right = self._insert(root.right, node)

        return root

    # INORDER (Left -> Root -> Right)
    def inorder(self, root):
        if root is None:
            return

        self.inorder(root.left)
        print(root.data, end=" ")
        self.inorder(root.right)

    # LEVEL ORDER (Level Wise)
    def level_order(self, root):
        if root is None:
            print("BST is Empty")
            return

        arr = []
        count = 0
        level = 1

        arr.append(root)

        while count < len(arr):
            level_size = len(arr) - count

            print("Level", level, ":", end=" ")

            for i in range(level_size):
                temp = arr[count]
                count += 1

                print(temp.data, end=" ")

                if temp.left:
                    arr.append(temp.left)

                if temp.right:
                    arr.append(temp.right)

            print()
            level += 1

# MAIN PROGRAM
b1 = BST()

b1.insert(Node(100))
b1.insert(Node(10))
b1.insert(Node(450))
b1.insert(Node(4))
b1.insert(Node(65))
b1.insert(Node(300))
b1.insert(Node(500))
b1.insert(Node(50))

# DISPLAY INORDER
print("Inorder Traversal:")
b1.inorder(b1.root)

# DISPLAY LEVEL ORDER
print("\nLevel Order Traversal:")
b1.level_order(b1.root)

print()
```
