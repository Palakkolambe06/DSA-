```python
# count node level wise 
# count how many on left or right 
class Node:
    def __init__(self, data):
        self.data = data
        self.left = None
        self.right = None

class BST:
    def __init__(self):
        self.root = None

    # INSERTION (Recursive)
    def insert(self, node):
        self.root = self._insert(self.root, node)

    def _insert(self, root, node):
        if root is None:
            return node
        if node.data < root.data:
            root.left = self._insert(root.left, node)
        elif node.data > root.data:
            root.right = self._insert(root.right, node)

        return root

    # FIND SUCCESSOR
    def get_successor(self, root):
        successor = root.right

        while successor.left is not None:
            successor = successor.left

        return successor

    # DELETION (Recursive)
    def del_node(self, value):
        self.root = self._delete(self.root, value)

    def _delete(self, root, value):
        if root is None:
            return None
        if value < root.data:
            root.left = self._delete(root.left, value)

        elif value > root.data:
            root.right = self._delete(root.right, value)

        else:
            # Zero or one child
            if root.left is None:
                return root.right

            if root.right is None:
                return root.left

            # Two children
            successor = self.get_successor(root)
            root.data = successor.data
            root.right = self._delete(
                root.right, successor.data
            )
      return root

    # INORDER: Left -> Root -> Right
    def inorder(self, root):
        if root is None:
            return

        self.inorder(root.left)
        print(root.data, end=" ")
        self.inorder(root.right)

    # PREORDER: Root -> Left -> Right
    def preorder(self, root):
        if root is None:
            return

        print(root.data, end=" ")
        self.preorder(root.left)
        self.preorder(root.right)

    # POSTORDER: Left -> Right -> Root
    def postorder(self, root):
        if root is None:
            return

        self.postorder(root.left)
        self.postorder(root.right)
        print(root.data, end=" ")

    # SEARCH (Key)
    def search(self, key):
        temp = self.root
        while temp is not None:
            if temp.data == key:
                return True

            elif key < temp.data:
                temp = temp.left

            else:
                temp = temp.right

        return False

    # MINIMUM VALUE
    def find_min(self, root):
        if root is None:
            return None

        if root.left is None:
            return root.data

        return self.find_min(root.left)

    # MAXIMUM VALUE
    def find_max(self, root):
        if root is None:
            return None

        if root.right is None:
            return root.data

        return self.find_max(root.right)

    # COUNT LEAF NODES (Recursive)
    def count_leaf(self, root):
        if root is None:
            return 0

        if root.left is None and root.right is None:
            return 1

        return (self.count_leaf(root.left) +
                self.count_leaf(root.right))

    # LEVEL ORDER TRAVERSAL
    def level_order(self):
        if self.root is None:
            return

        queue = [self.root]

        while queue:
            temp = queue.pop(0)
            print(temp.data, end=" ")

            if temp.left is not None:
                queue.append(temp.left)

            if temp.right is not None:
                queue.append(temp.right)


# MAIN PROGRAM
b1 = BST()

# INSERTION
b1.insert(Node(100))
b1.insert(Node(10))
b1.insert(Node(450))
b1.insert(Node(4))
b1.insert(Node(65))
b1.insert(Node(300))
b1.insert(Node(500))
b1.insert(Node(50))

# TRAVERSALS
print("Inorder:")
b1.inorder(b1.root)

print("\nPreorder:")
b1.preorder(b1.root)

print("\nPostorder:")
b1.postorder(b1.root)

print("\nLevel Order:")
b1.level_order()

# SEARCH
print("\n\nSearching 90:")
if b1.search(90):
    print("Found the node")
else:
    print("Node not found")

# MINIMUM AND MAXIMUM
print("\nMinimum:", b1.find_min(b1.root))
print("Maximum:", b1.find_max(b1.root))

# COUNT LEAF NODES
print("Total Leaf Nodes:", b1.count_leaf(b1.root))

# DELETION
b1.del_node(4)

print("\nInorder After Deletion:")
b1.inorder(b1.root)

# COUNT LEAF NODES AFTER DELETION
print("\nLeaf Nodes After Deletion:",
      b1.count_leaf(b1.root))

print()
```
